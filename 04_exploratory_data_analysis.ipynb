# Hotel Booking Cancellation & Overbooking Decision
## 04 — Exploratory Data Analysis (EDA)

This notebook explores cancellation patterns in the cleaned dataset. No machine-learning model is trained here.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', None)
```

## 2. Load Cleaned Dataset

```python
DATA_PATH='../data/processed/hotel_bookings_cleaned.csv'
df=pd.read_csv(DATA_PATH)
print('Shape:',df.shape)
df.head()
```

## 3. Overall Cancellation Distribution

```python
counts=df['is_canceled'].value_counts().sort_index()
print(counts)
print((df['is_canceled'].value_counts(normalize=True).sort_index()*100).round(2))
```

```python
counts.plot(kind='bar')
plt.title('Overall Booking Cancellation Distribution')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks([0,1],['Not Cancelled','Cancelled'],rotation=0)
plt.tight_layout(); plt.show()
```

## 4. Cancellation Rate by Hotel Type

```python
hotel_cancel=df.groupby('hotel')['is_canceled'].mean().mul(100).sort_values(ascending=False)
hotel_cancel
```

```python
hotel_cancel.plot(kind='bar')
plt.title('Cancellation Rate by Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout(); plt.show()
```

## 5. Cancellation Rate by Arrival Month

```python
month_order=['January','February','March','April','May','June','July','August','September','October','November','December']
monthly=df.groupby('arrival_date_month')['is_canceled'].mean().reindex(month_order).mul(100)
monthly
```

```python
monthly.plot(kind='bar',figsize=(10,5))
plt.title('Cancellation Rate by Arrival Month')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```

## 6. Lead Time and Cancellation

```python
df.groupby('is_canceled')['lead_time'].agg(['count','mean','median','std'])
```

```python
df.boxplot(column='lead_time',by='is_canceled',figsize=(8,5))
plt.title('Lead Time by Cancellation Status'); plt.suptitle('')
plt.xlabel('0 = Not Cancelled, 1 = Cancelled'); plt.ylabel('Lead Time (Days)')
plt.tight_layout(); plt.show()
```

## 7. Market Segment

```python
market=df.groupby('market_segment')['is_canceled'].agg(['count','mean'])
market['cancellation_rate']=market['mean']*100
market.sort_values('cancellation_rate',ascending=False)
```

```python
market['cancellation_rate'].sort_values(ascending=False).plot(kind='bar',figsize=(9,5))
plt.title('Cancellation Rate by Market Segment')
plt.ylabel('Cancellation Rate (%)'); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```

## 8. Deposit Type

```python
deposit=df.groupby('deposit_type')['is_canceled'].agg(['count','mean'])
deposit['cancellation_rate']=deposit['mean']*100
deposit.sort_values('cancellation_rate',ascending=False)
```

```python
deposit['cancellation_rate'].sort_values(ascending=False).plot(kind='bar')
plt.title('Cancellation Rate by Deposit Type')
plt.ylabel('Cancellation Rate (%)'); plt.xticks(rotation=0)
plt.tight_layout(); plt.show()
```

## 9. Previous Cancellations

```python
previous=df.groupby('previous_cancellations')['is_canceled'].agg(['count','mean'])
previous['cancellation_rate']=previous['mean']*100
previous.head(15)
```

## 10. Customer Type

```python
customer=df.groupby('customer_type')['is_canceled'].mean().mul(100).sort_values(ascending=False)
customer
```

```python
customer.plot(kind='bar')
plt.title('Cancellation Rate by Customer Type')
plt.ylabel('Cancellation Rate (%)'); plt.xticks(rotation=30)
plt.tight_layout(); plt.show()
```

## 11. ADR and Cancellation

```python
df.groupby('is_canceled')['adr'].agg(['count','mean','median','std'])
```

```python
df.boxplot(column='adr',by='is_canceled',figsize=(8,5))
plt.title('ADR by Cancellation Status'); plt.suptitle('')
plt.xlabel('0 = Not Cancelled, 1 = Cancelled'); plt.ylabel('Average Daily Rate')
plt.tight_layout(); plt.show()
```

## 12. Distribution Channel

```python
channel=df.groupby('distribution_channel')['is_canceled'].agg(['count','mean'])
channel['cancellation_rate']=channel['mean']*100
channel.sort_values('cancellation_rate',ascending=False)
```

```python
channel['cancellation_rate'].sort_values(ascending=False).plot(kind='bar')
plt.title('Cancellation Rate by Distribution Channel')
plt.ylabel('Cancellation Rate (%)'); plt.xticks(rotation=30)
plt.tight_layout(); plt.show()
```

## 13. New vs Repeated Guests

```python
repeat=df.groupby('is_repeated_guest')['is_canceled'].agg(['count','mean'])
repeat['cancellation_rate']=repeat['mean']*100
repeat
```

```python
repeat['cancellation_rate'].plot(kind='bar')
plt.title('Cancellation Rate: New vs Repeated Guests')
plt.ylabel('Cancellation Rate (%)')
plt.xticks([0,1],['New Guest','Repeated Guest'],rotation=0)
plt.tight_layout(); plt.show()
```

## 14. Numerical Correlation with Cancellation

```python
numeric=df.select_dtypes(include=np.number)
corr=numeric.corr()['is_canceled'].drop('is_canceled').sort_values()
corr
```

```python
corr.plot(kind='barh',figsize=(9,9))
plt.title('Numerical Feature Correlation with Cancellation')
plt.xlabel('Pearson Correlation with is_canceled')
plt.tight_layout(); plt.show()
```

### Note
Correlation measures association, not causation. Categorical variables are not fully represented in this numerical correlation view.

## 15. Key EDA Results

```python
print('Overall cancellation rate:',round(df['is_canceled'].mean()*100,2),'%')
print('\nBy hotel:')
print(hotel_cancel.round(2))
print('\nTop market segments:')
print(market['cancellation_rate'].sort_values(ascending=False).round(2).head())
print('\nBy deposit type:')
print(deposit['cancellation_rate'].sort_values(ascending=False).round(2))
```

## EDA Summary

This notebook studies cancellation associations across hotel type, season, lead time, market segment, deposit type, cancellation history, customer type, ADR, distribution channel and repeated-guest status.
